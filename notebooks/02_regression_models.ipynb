```python
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

# Obtener los datos
x = pima['processor_gnrtn']
y = pima['Price']

# Crear el modelo de regresión lineal
mod0 = smf.ols('y ~ x', data=pima).fit()

# Obtener los coeficientes del modelo
b0 = mod0.params[0]
b1 = mod0.params[1]

# Graficar los datos y la línea de regresión
plt.figure(figsize=(8, 6))
plt.scatter(x, y, s=50)
plt.plot(x, b0 + b1*x, color='red', linewidth=2)

# Agregar etiquetas y título
plt.xlabel('processor_gnrtn')
plt.ylabel('Price')
plt.title('Regresión Lineal')

plt.show()

# Mostrar la ecuación de la regresión
equation = f'Price = {b0:.2f} + {b1:.2f} * x'
print("La ecuación de la regresión es:", equation)

# Obtener el valor de r^2
r_squared = mod0.rsquared
print("El valor de r^2 para la regresión lineal es:", r_squared)
```

**Procedimiento:**

Este código realiza una regresión lineal simple entre las variables 'processor_gnrtn' (eje x) y 'Price' (eje y) de uel conjunto de datos Pima. Primero, crea un modelo de regresión lineal utilizando el método de mínimos cuadrados ordinarios (OLS) de statsmodels. Luego, ajusta el modelo a los datos y extrae los coeficientes de la línea de regresión. Después, genera un gráfico que muestra los puntos de datos y la línea de regresión. Añade etiquetas a los ejes y un título al gráfico. Finalmente, muestra la ecuación de la línea de regresión y el valor de r^2, que representa la proporción de la varianza en la variable dependiente que es predecible a partir de la variable independiente.

**Conclusiones:**

Viendo los resulados obtenidos, la linea de tendecia en esta regresion lineal se puede definir como Price = 104799.12 + -2696.77 * x, la cual tiene un 0.002 de coeficiencia osea un 0.2% de relación, lo cual es extremadamente baja y nos muestra que para obtener una mejor relación, se debe realizar otro tipo de relación que se NO lineal.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from scipy.optimize import curve_fit

# Obtener los datos
x = np.array(pima['processor_gnrtn'])
y = np.array(pima['Price'])

# Crear el modelo de regresión exponencial
def exponential_func(x, a, b):
    return a * np.exp(b * x)

# Ajustar el modelo a los datos originales
popt, pcov = curve_fit(exponential_func, x, y)

# Parámetros del ajuste
a, b = popt

# Graficar los datos originales y la curva ajustada
plt.scatter(x, y, label='Datos')
plt.plot(x, exponential_func(x, a, b), 'r--', label='Regresión Exponencial')
plt.xlabel('processor_gnrtn')
plt.ylabel('Price')
plt.legend()
plt.title('Regresión Exponencial')
plt.show()

# Calcular el valor de r^2
r2 = r2_score(y, exponential_func(x, a, b))
print("El valor de r^2 al cuadrado para la regresión exponencial es:", r2)

# Imprimir la ecuación de la regresión exponencial
print(f"Ecuación de la regresión exponencial: Price = {a:.2f} * e^({b:.2f} * x)")
```

Este código realiza una regresión exponencial sobre los datos de `processor_gnrtn` (eje x) y `Price` (eje y) del conjunto de datos `pima`. Primero, define una función exponencial llamada `exponential_func` con parámetros `a` y `b`, que representan los coeficientes de la regresión exponencial. Luego, ajusta el modelo exponencial a los datos originales utilizando el método `curve_fit` de la biblioteca `scipy.optimize`. Después de obtener los parámetros del ajuste, grafica los datos originales y la curva de regresión exponencial. Calcula el valor de r^2 para evaluar la bondad del ajuste. **Conclusiones:**
