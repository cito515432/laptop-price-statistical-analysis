##La Distribución Binomial

<p>$$f\left( x \right)=P\left( X=x \right)=\binom{n}{x}p^x(1-p)^{n-x}$$</p>
<p>v.a. $X$: No. de éxitos en $n$ ensayos</p>
<p><strong>Parámetros:</strong><br> $p$ : probabilidad de éxito, <br>
$n$: Número de ensayos</p><br>
<p><strong>Primeros momentos:</strong>
<p>Media: $\mu = n*p$</p>
<p>Varianza: $\sigma^2=n*p*(1-p)$</p>

##Ejercicio 1.
<p>Al probar cierta clase de neumático para camión en un terreno accidentado, se encuentra que el 25% de los camiones no completan la prueba de recorrido sin pinchaduras. De los siguientes 15 camiones probados, calcule la probabilidad de que:<br>
a) de 3 a 6 tengan pinchaduras (Realice la suma con probabilidades puntuales o exactas y también con la distribución acumulada)<br>
b) menos de 4 tengan pinchaduras: $P(X<4)=P(X<=3)$. Realice la suma con probabilidades puntuales o exactas y también con la distribución acumulada.<br>
c) al menos 5 presenten pinchaduras.<br>
d) a lo sumo 6 presenten pinchaduras.



```python
#Solución
from math import comb, factorial
#Forma manual por Wilson Castro Z.:
#Se calculan las tablas de probabilidades del ejercicio y se tabulan:
n = 15
p = 0.25
print('Distribucion de probabilidad binomial')
PXAcum=0
print('P(X = x) \t\tP(X <= x)')
for i in range(n+1):
  PX=round(comb(n,i)*p**i*(1-p)**(n-i),4)
  PXAcum+=PX
  PXAcum=round(PXAcum,4)
  print('P(X = ',i,') =',PX,"\t",PXAcum)
#Se soluciona el ejercicio:
def binomial(n, k, p):
   """
   Función para calcular la probabilidad binomial.
   n: número de ensayos
   k: número de éxitos
   p: probabilidad de éxito
   """
   return comb(n, k) * p**k * (1-p)**(n-k)

n = 15  # Total de camiones
p = 0.25  # Probabilidad de pinchadura

# a) Probabilidad de que de 3 a 6 camiones tengan pinchaduras
p_3_a_6 = sum(binomial(n, k, p) for k in range(3, 7))
print(f"a) Probabilidad de que de 3 a 6 camiones tengan pinchaduras: {p_3_a_6:.4f}")

# b) Probabilidad de que menos de 4 camiones tengan pinchaduras
p_menos_4 = sum(binomial(n, k, p) for k in range(4))
print(f"b) Probabilidad de que menos de 4 camiones tengan pinchaduras: {p_menos_4:.4f}")

# c) Probabilidad de que más de 5 camiones tengan pinchaduras
p_mas_5 = sum(binomial(n, k, p) for k in range(6, n+1))
print(f"c) Probabilidad de que más de 5 camiones tengan pinchaduras: {p_mas_5:.4f}")
```

##Ejercicio 2.
La probabilidad de que un paciente se recupere de una delicada operación de corazón es 0.9. ¿Cuál es la probabilidad de que exactamente 5 de los siguientes
7 pacientes intervenidos sobrevivan?

```python
#Solución:
from math import comb
#Forma manual por Wilson Castro Z.:
#Se calculan las tablas de probabilidades del ejercicio y se tabulan:
n = 7
p = 0.9
print('Distribucion de probabilidad binomial')
PXAcum=0
print('P(X = x) \t\tP(X <= x)')
for i in range(n+1):
  PX=round(comb(n,i)*p**i*(1-p)**(n-i),4)
  PXAcum+=PX
  PXAcum=round(PXAcum,4)
  print('P(X = ',i,') =',PX,"\t",PXAcum)
#Se soluciona el ejercicio:
def binomial(n, k, p):
    """
    Calcula la probabilidad binomial de obtener k éxitos en n ensayos
    con una probabilidad de éxito p.
    """
    return comb(n, k) * p**k * (1-p)**(n-k)

# Parámetros
n = 7  # Número total de pacientes
k = 5  # Número de pacientes que sobreviven
p = 0.9  # Probabilidad de que un paciente se recupere

# Calcular la probabilidad de que exactamente 5 de los 7 pacientes sobrevivan
probabilidad = binomial(n, k, p)
print(f"La probabilidad de que exactamente 5 de los 7 pacientes sobrevivan es: {probabilidad:.4f}")
```

##Ejercicio 3.

b) menos de 2 contraigan la enfermedad.<br>

```python
#Solución
from math import comb
#Forma manual por Wilson Castro Z.:
#Se calculan las tablas de probabilidades del ejercicio y se tabulan:
n = 5
p = 0.6
print('Distribucion de probabilidad binomial')
PXAcum=0
print('P(X = x) \t\tP(X <= x)')
for i in range(n+1):
  PX=round(comb(n,i)*p**i*(1-p)**(n-i),4)
  PXAcum+=PX
  PXAcum=round(PXAcum,4)
  print('P(X = ',i,') =',PX,"\t",PXAcum)
#Se soluciona el ejercicio:
def binomial(n, k, p):
    """
    Calcula la probabilidad binomial de obtener k éxitos en n ensayos
    con una probabilidad de éxito p.
    """
    return comb(n, k) * p**k * (1-p)**(n-k)

# Parámetros
n = 5  # Número total de ratones
p = 0.6  # Probabilidad de que un ratón quede protegido

# a) Probabilidad de que ninguno contraiga la enfermedad
k = n  # Todos los ratones quedan protegidos
prob_ninguno = binomial(n, k, p)
print(f"a) Probabilidad de que ninguno contraiga la enfermedad: {prob_ninguno:.4f}")

# b) Probabilidad de que menos de 2 contraigan la enfermedad
prob_menos_2 = sum(binomial(n, k, p) for k in range(3, n+1))
print(f"b) Probabilidad de que menos de 2 contraigan la enfermedad: {prob_menos_2:.4f}")

# c) Probabilidad de que más de 3 contraigan la enfermedad
prob_mas_3 = sum(binomial(n, k, p) for k in range(4, n+1))
print(f"c) Probabilidad de que más de 3 contraigan la enfermedad: {prob_mas_3:.4f}")

# d) Probabilidad de que entre 2 y 3 ratones se enfermen
prob_2_3 = binomial(n, 2, p) + binomial(n, 3, p)
print(f"d) Probabilidad de que entre 2 y 3 ratones se enfermen: {prob_2_3:.4f}")
```

#DISTRIBUCION DE POISSON
<p>En esta distribución, la función de masa de probabilidad es:</p>
<p>$$f(x) = P( X=x) = \frac{{e^{ - \lambda } \lambda ^x }}{{x!}}$$</p>
<p>Con lambda -$\lambda$- como la media de éxitos que ocurren por unidad de área, intervalo de tiempo, por pieza, etc. y la variable aleatoria X como el número de éxitos u ocurrencias en los mismos términos que $\lambda$.</p>

<p>Media: $\mu = \lambda$</p>
<p>Varianza: $\sigma^2 = \lambda$</p>


## Ejercicio 4

<p>Un escritor de libros comete, en promedio, dos errores de procesamiento de texto por
página en el primer borrador de su libro. ¿Cuál es la probabilidad de que en la siguiente
página cometa:<br>
a) ¿4 o más errores?<br>
b) ¿Ningún error?</p>

```python
from math import exp, factorial

# Función para calcular la probabilidad de Poisson
def poisson(k, lam):
    return (exp(-lam) * lam**k) / factorial(k)

# Crear la tabla de probabilidades de Poisson
n = 4
mu = 2
print('Distribución de probabilidad de Poisson')
PXAcum = 0
print('P(X = x) \t\tP(X <= x)')
for i in range(n+1):
    PX = poisson(i, mu)
    PXAcum += PX
    PX = round(PX, 4)
    PXAcum = round(PXAcum, 4)
    print('P(X = ', i, ') =', PX, "\t", PXAcum)

# Media de la distribución de Poisson
lam = 2  # Dos errores en promedio por página

# Parte a: Probabilidad de cometer 4 o más errores en la siguiente página
prob_a = 1 - sum(poisson(k, lam) for k in range(4))
print(f"a) Probabilidad de cometer 4 o más errores: {prob_a:.6f}")

# Parte b: Probabilidad de no cometer ningún error
prob_b = poisson(0, lam)
print(f"b) Probabilidad de no cometer ningún error: {prob_b:.6f}")
```

```python
from math import exp, factorial
# Crear la tabla de probabilidades de Poisson
n = 3
mu = 5
print('Distribucion de probabilidad de Poisson')
PXAcum=0
print('P(X = x) \t\tP(X <= x)')
for i in range(n+1):
  PX=poisson(i,mu)
  PXAcum+=PX
  PX=round(PX,4)
  PXAcum=round(PXAcum,4)
  print('P(X = ',i,') =',PX,"\t",PXAcum)

# Media de la distribución de Poisson
lam = 5

# Función para calcular la probabilidad de Poisson
def poisson(k, lam):
    return (exp(-lam) * lam**k) / factorial(k)

# Parte a: Proporción de días en los que la demanda por un ejemplar es cero
prob_a = poisson(0, lam)
print(f"a) Proporción de días en los que la demanda por un ejemplar es cero: {prob_a:.6f}")

# Parte b: Proporción de días en los que la demanda por un ejemplar del libro supera la oferta
prob_b = 1 - sum(poisson(k, lam) for k in range(4))
print(f"b) Proporción de días en los que la demanda por un ejemplar del libro supera la oferta: {prob_b:.6f}")

# Parte c: Probabilidad de que se presenten 21 solicitudes durante el evento especial de "los tres días de la lectura"
prob_c = poisson(21, 3 * lam)
print(f"c) Probabilidad de que se presenten 21 solicitudes durante el evento especial: {prob_c:.6f}")
```

## Ejercicio 6. Un fabricante de automóviles se preocupa por una falla en el mecanismo de freno de
un modelo específico. En raras ocasiones la falla puede causar una catástrofe al
manejarlo a alta velocidad. La distribución del número de automóviles por año que
experimentará la catástrofe es una variable aleatoria de Poisson con λt=5.
a) ¿Cuál es la probabilidad de que, a lo sumo, 3 automóviles por año de ese
modelo específico sufran una catástrofe?
b) ¿Cuál es la probabilidad de que más de un automóvil por año experimente una
catástrofe?

```python
from math import exp, factorial

# Crear la tabla de probabilidades de Poisson
n = 3
mu = 5
print('Distribucion de probabilidad de Poisson')
PXAcum=0
print('P(X = x) \t\tP(X <= x)')
for i in range(n+1):
  PX=poisson(i,mu)
  PXAcum+=PX
  PX=round(PX,4)
  PXAcum=round(PXAcum,4)
  print('P(X = ',i,') =',PX,"\t",PXAcum)

# Media de la distribución de Poisson
lam = 5

# Función para calcular la probabilidad de Poisson
def poisson(k, lam):
    return (exp(-lam) * lam**k) / factorial(k)

# Parte a: Probabilidad de que, a lo sumo, 3 automóviles por año de ese modelo específico sufran una catástrofe
prob_a = sum(poisson(k, lam) for k in range(4))
print(f"a) Probabilidad de que, a lo sumo, 3 automóviles por año sufran una catástrofe: {prob_a:.6f}")

# Parte b: Probabilidad de que más de un automóvil por año experimente una catástrofe
prob_b = 1 - poisson(0, lam) - poisson(1, lam)
print(f"b) Probabilidad de que más de un automóvil por año experimente una catástrofe: {prob_b:.6f}")
```

7. Un investigador informa que unos ratones a los que primero se les restringen drásticamente sus dietas y después se les enriquecen con vitaminas y proteínas vivirán un promedio de 40 meses. Si suponemos que la vida de tales ratones se distribuye
normalmente, con una desviación estándar de 6.3 meses, calcule la probabilidad de que un ratón determinado viva:

a) más de 32 meses;<br>
b) menos de 28 meses;<br>
c) entre 37 y 49 meses.<br>

```python
import math
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

def normal_pdf(x, mu, sigma):
   return (1 / (sigma * math.sqrt(2 * math.pi))) * math.exp(-0.5 * ((x - mu) / sigma) ** 2)

def plot_normal_dist(mu, sigma):
   x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
   y = [normal_pdf(x_i, mu, sigma) for x_i in x]
   plt.figure(figsize=(10, 6))
   plt.plot(x, y, 'r-', lw=2, label='Distribución Normal')
   plt.xlabel('Duración de vida (meses)')
   plt.ylabel('Densidad de Probabilidad')
   plt.title(f'Distribución Normal con $\mu$ = {mu}, $\sigma$ = {sigma}')
   plt.legend()
   plt.show()

# Datos del problema
mu = 40  # Media de la vida de los ratones
sigma = 6.3  # Desviación estándar

plot_normal_dist(mu, sigma)

# a) Probabilidad de que un ratón viva más de 32 meses
a = 1 - norm.cdf(32, mu, sigma)
print(f"a) Probabilidad de vivir más de 32 meses: {a:.4f}")

# b) Probabilidad de que un ratón viva menos de 28 meses
b = norm.cdf(28, mu, sigma)
print(f"b) Probabilidad de vivir menos de 28 meses: {b:.4f}")

# c) Probabilidad de que un ratón viva entre 37 y 49 meses
c = norm.cdf(49, mu, sigma) - norm.cdf(37, mu, sigma)
print(f"c) Probabilidad de vivir entre 37 y 49 meses: {c:.4f}")
```

9. La vida promedio de cierto tipo de motor pequeño es de 10 años, con una desviación estándar de 2 años. El fabricante reemplaza gratis todos los motores que fallen dentro del periodo de garantía. Si estuviera dispuesto a reemplazar sólo 3% de los motores que fallan, ¿cuánto tiempo de garantía debería ofrecer? Suponga que la duración de un
motor sigue una distribución normal.

```python
import math
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

def normal_pdf(x, mu, sigma):
   return (1 / (sigma * math.sqrt(2 * math.pi))) * math.exp(-0.5 * ((x - mu) / sigma) ** 2)

def plot_normal_dist(mu, sigma, tiempo_garantia):
   x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
   y = [normal_pdf(x_i, mu, sigma) for x_i in x]
   plt.figure(figsize=(10, 6))
   plt.plot(x, y, 'r-', lw=2, label='Distribución Normal')
   plt.xlabel('Duración del motor (años)')
   plt.ylabel('Densidad de Probabilidad')
   plt.title(f'Distribución Normal con $\mu$ = {mu}, $\sigma$ = {sigma}')
   plt.axvline(x=tiempo_garantia, color='g', linestyle='--', label=f'Tiempo de garantía: {tiempo_garantia:.2f} años')
   plt.legend()
   plt.show()

# Parámetros de la distribución normal
mu = 10  # Media
sigma = 2  # Desviación estándar

# Calcular el tiempo de garantía para reemplazar solo el 3% de los motores que fallan
porcentaje_reemplazar = 0.03  # 3%
tiempo_garantia = norm.ppf(1 - porcentaje_reemplazar, mu, sigma)

plot_normal_dist(mu, sigma, tiempo_garantia)

print(f"El fabricante debería ofrecer un tiempo de garantía de {tiempo_garantia:.2f} años para reemplazar solo el 3% de los motores que fallan.")
```
