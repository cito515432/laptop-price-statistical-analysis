Estadisticos de la variable Precios en Euros y Diagrama de tallos y hojas:

```python
import numpy as np

important_features = ['laptop_ID', 'Company', 'Product', 'TypeName', 'Inches', 'ScreenResolution', 'Cpu', 'Ram', 'Memory', 'Gpu', 'OpSys', 'Weight', 'Price_euros']
datos_data = datos.loc[:, important_features]
print('Estadísticos sobre la variable "Price_euros"')
def medidas_posicion(columna):
    media = np.mean(columna)
    mediana = np.median(columna)
    percentil_25 = np.percentile(columna, 25)
    percentil_75 = np.percentile(columna, 75)

    return media, mediana, percentil_25, percentil_75

def diagrama_tallos_hojas(columna):

    tallos = [int(x // 10) for x in columna]
    hojas = [int(x % 10) for x in columna]

    diagrama = {}
    for t, h in zip(tallos, hojas):
        if t not in diagrama:
            diagrama[t] = []
        diagrama[t].append(h)

    for t, h in sorted(diagrama.items()):
        print(f'{t} | {" ".join(map(str, h))}')


datos1 = np.array(datos_data['Price_euros'])
media, mediana, p25, p75 = medidas_posicion(datos_data['Price_euros'])

print(f'Media: {media}')
print(f'Mediana: {mediana}')
print(f'Percentil 25: {p25}')
print(f'Percentil 75: {p75}')

print('\nDiagrama de Tallos y Hojas:')
diagrama_tallos_hojas(datos1)
```

**Procedimiento:**
Importamos nuevamente la misma librería “import numpy as np”, nombramos nuevamente nuestra lista “important_features” e imprimimos “'Estadísticos sobre la variable "Price_euros””. Terminando esta parte usamos “medidas_posicion(columna)” donde nos ayudara a encontrar la media, mediana y el cuartil superior e inferior para cada columna, estos resultados los pondremos en variables.
Ahora definimos una función llamada “diagrasma_tallos_hojas” esto creara un diagrama de tallos y hojas para la distribución de cada dato en una columna dada. En “tallos = [int(x // 10) for x in columna]” estamos creando los tallos van a ser cada valor de la columna entre 10 siendo esta una división entera por eso los dobles // tomando la parte entera del resultado, proporcionando asi los tallos del diagrama. Para las hojas se ejecuta “hojas = [int(x % 10) for x in columna]” lo que calculara el residuo de la división entre 10 por cada valor en la columna dando así las hojas del diagrama. Teniendo ya los tallos y las hojas creamos la variable “diagrama” que almacenara estos datos en una lista, el bucle que estamos creando por medio de los tallos (t) y las hojas (h)  y la función “zip” asignaremos las hojas a los tallos correspondientes en el diccionario, en donde si el tallo no está presente el diccionario creara una nueva entrada y consigo una lista vacía para las hojas, para finalizar esta parte se imprimirá el diagrama ordenado por los tallos, mostrando los tallos seguidos de las hojas que corresponda.
Para la última parte del código creemos un array de numpy con los datos de los precios, como también llamaremos la media, mediana y los dos cuartiles para ser calculados y finalmente imprimimos el diagrama de tallos y hojas.

```python
data=datos_data['Company']
n=data.value_counts(sort=False)
N=n.cumsum()
f=100*data.value_counts(sort=False)/len(data)
F=100*f.cumsum()/f.sum()
```

**Procedimientos:**

Llamamos a nuestro “datos_data” con la columna de “Company” ahora teniendo nuestros datos, usaremos la siguiente línea “n=data.value_counts(sort=False)”  esta línea cuenta la frecuencia de cada una de los datos de “Company”  y los almacena en la serie llamada “n” con un parámetro “sort=False” evita que los valores se ordenen por frecuencia. Ahora “N=n.cumsum()” acá calcula las frecuencias acumuladas. Con la siguiente línea de código “f=100data.value_counts(sort=False)/len(data)” calculara la frecuencia relativa de cada único valor de la serie “data” donde esta se calcula dividiendo la frecuencia de cada valor por el numero total de elementos de “data” y se multiplican por 100 de esta forma se obtiene un porcentaje. Estos datos se guardarán en f. Y en la ultima parte de este código “F=100f.cumsum()/f.sum()” que es similar a la línea anterior la diferencia es que se van a calcular las frecuencias relativas acumuladas se utilizara primero “cumsun()” que hará la sumara acumulativa de las frecuencias de nuestra variable f, luego estas se divide por la suma total de frecuencias relativas y se multiplica por 100 para obtener un porcentaje acumulado. Esto se guardará en F.
