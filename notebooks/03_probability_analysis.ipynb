# tabla de distribución de probabilidad

***Primer analisis: Precio en Euros***

```python
import pandas as pd
import numpy as np

# Definir el número de intervalos
num_bins = 10

# Crear los intervalos (bins)
bins = np.linspace(datos_data['Price_euros'].min(), datos_data['Price_euros'].max(), num_bins + 1)

# Calcular la frecuencia de datos en cada intervalo
counts, edges = np.histogram(datos_data['Price_euros'], bins=bins)

# Calcular la probabilidad para cada intervalo
probabilities = counts / counts.sum()

# Calcular las probabilidades acumuladas
cumulative_probabilities = np.cumsum(probabilities)

# Crear un DataFrame para la tabla de distribución de probabilidad
dist_prob_table = pd.DataFrame({
    'Interval': pd.IntervalIndex.from_breaks(edges),
    'Frequency': counts,
    'P(X=x)': probabilities,
    'P(X<=x)': cumulative_probabilities
})
print("\nDistribución de probabilidad de Los precios en Euros:")
print("r                                         tp(r)")
# Mostrar la tabla con el formato solicitado
for i, row in dist_prob_table.iterrows():
    print(f"P(X in {row['Interval']}) = {row['P(X=x)']:.4f} \t {row['P(X<=x)']:.4f}")

# Calcular la media y la varianza de los datos
mean_price = datos_data['Price_euros'].mean()
var_price = datos_data['Price_euros'].var()

# Mostrar la media y la varianza
print(f"\nMedia: {mean_price:.2f}")
print(f"Varianza: {var_price:.2f}")
```

**Procedimiento:**
Definir Intervalos (Bins): np.linspace genera intervalos equiespaciados basados en el mínimo y máximo de Price_euros.
Calcular Frecuencia: np.histogram cuenta la cantidad de datos que caen en cada intervalo.
Calcular Probabilidad: Se divide la frecuencia de cada intervalo por la suma total de frecuencias.
Calcular Probabilidades Acumuladas: np.cumsum calcula la suma acumulativa de las probabilidades.
Crear DataFrame: Se crea un DataFrame con los intervalos, la frecuencia, la probabilidad puntual (P(X=x)) y la probabilidad acumulada (P(X<=x)).
Mostrar la Tabla: Se imprime la tabla de distribución de probabilidad en el formato solicitado.
Calcular Media y Varianza: Se calculan la media y la varianza de los datos de Price_euros.
Mostrar Media y Varianza: Se imprimen la media y la varianza.
Al final se genera una tabla de distribución de probabilidad para la variable Price_euros y mostrará las probabilidades puntuales y acumuladas de cada intervalo, junto con la media y la varianza al final.

 La mayoría de los precios se encuentran en los primeros dos intervalos. El 36.38% de los precios están en el rango de (174.0, 766.5] y el 34.38% están en el rango de (766.5, 1359.0]. Esto significa que aproximadamente el 70.76% de los precios están por debajo de 1359 euros.

**Distribuición teórica que más se ajusta:**
Dado que la variable aleatoria es continua y positiva, y la distribución de probabilidad parece tener una forma asimétrica hacia la derecha, una distribución teórica candidata sencilla podría ser la distribución exponencial.
La distribución exponencial se utiliza comúnmente para modelar el tiempo entre eventos en un proceso de Poisson, pero también se puede aplicar a otras variables aleatorias continuas positivas con una tasa de decaimiento constante.
Para calcular el valor esperado (media) de una variable aleatoria que sigue una distribución exponencial en Python, podemos utilizar la función expon.mean() de la biblioteca scipy.stats.

```python
import numpy as np
from scipy.stats import expon

# Datos de precios en euros
precios_euros = datos_data['Price_euros']

# Ajustar la distribución exponencial a los datos
parametro_lambda = 1 / precios_euros.mean()  # Estimación del parámetro lambda
dist_exp = expon(scale=1/parametro_lambda)

# Calcular el valor esperado (media) de la distribución exponencial
valor_esperado = dist_exp.mean()

print(f"Valor esperado (media) de la distribución exponencial ajustada: {valor_esperado:.2f}")
```

**Procedimiento:**
Estimamos el parámetro lambda de la distribución exponencial utilizando la inversa de la media de los precios en euros.
Creamos una instancia de la distribución exponencial dist_exp con el parámetro scale=1/parametro_lambda.
Calculamos el valor esperado (media) de la distribución exponencial utilizando dist_exp.mean().
Finalmente, imprimimos el valor esperado con dos decimales.

**Conclusiones:**
La media de los datos de precios en euros (1123.69) coincide con el valor esperado de la distribución exponencial ajustada. Esto sugiere que la distribución exponencial podría ser una aproximación razonable para modelar la distribución de los precios en euros.
La distribución exponencial es una distribución continua y positiva, lo que se ajusta a la naturaleza de los datos de precios, que son valores positivos.
La distribución exponencial tiene una forma asimétrica positiva, lo que también coincide con la observación de que la distribución de probabilidad de los precios en euros está sesgada hacia la derecha (asimetría positiva).

## Prueba Chi cuadrado ($\chi^2$) para verificar si los datos se distribuyen como la Poisson

R.A. Fisher mostró que bajo el supuesto de que los datos $X_1, X_2,...,X_n$ son generados por una distribución de Poisson con algún parámetro $\lambda$ (o bien $\mu$), entonces el estadístico:

$D =  \frac{ \sum_{i=1}^n (X_i- \overline{X})^2}{\overline{X}}$

donde $\overline{X}$ es la media muestral y tiene una $\chi^2$ con $n-1$ grados de libertad. El p-valor de la cola superior está dado por:

$p = 1 - F(x)$ donde $F(x)$ es la distribución acumulada, la cual en Python se toma con la librería stats de scipy (python científico) como  chi2.cdf (chi cuadrada de la cumulative distribution function), como se observa a continuación:

```python
# Código elaborado por Ing. Wilson Castro Z.
from scipy import stats
Dd = []
mu = np.mean(datos_data['Inches'])
for i in datos_data['Inches']:
  Dd.append((i-mu)**2)
D = sum(Dd)/mu
df = len(datos_data['Inches'])-1 # parametro de forma.
pvalor = 1-stats.chi2.cdf(D,df)
print(f'D = {D}\ndf = {df}\np-valor = {round(pvalor,3)}')
# Comparar el p-valor con un nivel de significancia (por ejemplo, 0.05)
alfa = 0.05
if pvalor < alfa:
    print("Se rechaza la hipótesis nula. Los datos no se ajustan a la distribución de Poisson.")
else:
    print("No se rechaza la hipótesis nula. Los datos se ajustan a la distribución de Poisson.")
```

**Conclusión:**

como el $p-value = 1.0 \geq 0.05$ -Tomando este último valor como el nivel de significancia ($\alpha = 0.05$), concluimos que no hay evidencia en contra de la hipótesis nula de que la verdadera distribución es de Poisson con una media de alrededor de 15.02 Pulgadas.

Teniendo en cuenta esto, realizamos la Distribuición de Poisson:

```python
import numpy as np
from scipy.stats import poisson

# Datos de tamaños en pulgadas
tamanos_pulgadas = datos_data['Inches']

# Estimar el parámetro lambda de la distribución de Poisson
lambda_estimado = tamanos_pulgadas.mean()

# Crear una instancia de la distribución de Poisson
dist_poisson = poisson(mu=lambda_estimado)

# Calcular el valor esperado de la distribución de Poisson
valor_esperado = dist_poisson.mean()

print(f"Valor esperado (media) de la distribución de Poisson ajustada: {valor_esperado:.2f}")
```
